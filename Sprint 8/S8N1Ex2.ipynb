```python
import pandas as pd
```

```python
df = pd.read_excel('Datos/sprint8_complex.xlsx', skiprows=3)
df.rename(columns={'Unnamed: 0':'Id'}, inplace=True)
```

```python
# ¿Existen identificadores únicos (DNI) para cada persona?
# No, pues algunos están duplicados.
# En el caso del DNI '90585882A'.
#   Id 104, formato de salario inadecuado.
#   Id 114, número de hijos inválido.
#   Id 281, formato de salario inadecuado.
#   Id 25, 913 y 1000, datos duplicados excepto en campo salario. Conservado el último considerando que son los datos más actualizados.
# En el caso del DNI '55283701C'.
#   Id 36 y 40, datos duplicados. Por defecto se mantiene el primero.
# En el caso del DNI '54013571W'.
#   Id 142, valores contradictorios en 'Fills' y 'No Fills'.
# En el caso del DNI '15909601K'.
#   Id 250, formato de salario inadecuado, valores contradictorios en 'Fills' y 'No Fills'.
# En el caso del DNI '97954019T'.
#   Id 372 y 1002, datos duplicados. Por defecto se mantiene el primero.
# En el caso del DNI '49529438G'.
#   Id 692, valor de salario mensual sin coma de millar, número de hijos inválido.
# En el caso del DNI '16618423V'.
#   Id 654, 754, 758 y 759, presentan inconsistencias. El más adecuado es el 758, que únicamente presenta un formato de salario inadecuado.
# En el caso del DNI '55297384F'.
#   Id 730 y 1003, datos duplicados. Por defecto se mantiene el primero.
# En el caso del DNI '52590001R'.
#   Id 757 y 1004, datos duplicados. Por defecto se mantiene el primero.

dupes_list = []
for i, dupe in enumerate(df['DNI'].duplicated(keep=False)):
    if dupe == True:
        dupes_list.append(df.iloc[i])
        # display(df.iloc[i])
dupes_df = pd.DataFrame(dupes_list)

display(dupes_df.sort_values('DNI'))
```

```python
# ¿Qué harás con las filas duplicadas o registros repetidos?
# Eliminar los duplicados siguiendo el criterio explicado anteriormente. 
pre_drop = len(df)
df.drop_duplicates(subset=df.columns.difference(['Id']), inplace=True)
post_drop = len(df)

diff = pre_drop - post_drop

display(df)
print(str(diff)+' registros eliminados.')
```

```python
# ¿Hay valores obligatorios que faltan o información incoherente?
# Sí.

df.dropna(subset=df.columns.difference(['Fills', 'No Fills', 'Km_anuals', 'Consum_mitja_L_100km']), inplace=True)
df.dropna(how='all', subset=['Fills', 'No Fills'], inplace=True)

for indx, [fills, no_fills] in enumerate(zip(df['Fills'], df['No Fills'])):
    if fills == no_fills:
        print(indx)
        df.drop(index=indx, errors='ignore')

display(df)
```
