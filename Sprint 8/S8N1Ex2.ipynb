```python
def repair_mojibake(texto):
    if isinstance(texto, str):
        try:
            return texto.encode('windows-1252').decode('utf-8')
        except (UnicodeEncodeError, UnicodeDecodeError):
            return texto
    return texto

nombres_columnas = [repair_mojibake(col) for col in df.columns]
df.columns = nombres_columnas
for columna in df:
    df[columna] = df[columna].apply(repair_mojibake)

df.head(10)
```
