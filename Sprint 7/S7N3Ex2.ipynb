```python
equipos = {}

def define_resultado(goles):
    if goles[0] > goles[1]:
        return ['victorias','derrotas']
    if goles[0] < goles[1]:
        return ['derrotas','victorias']
    if goles[0] == goles[1]:
        return ['empates','empates']
    
def calcula_puntos(resultados):
    puntos = [0, 0]
    for i in range(2):
        if resultados[i] == 'victorias':
            puntos[i] = 3
        elif resultados[i] == 'empates':
            puntos[i] = 1
        elif resultados[i] == 'derrotas':
            puntos[i] = 0
    return puntos

def actualiza_equipo(nombres_equipos, resultado_goles, resultados, puntos):
        equipos[nombres_equipos[0]]['goles_favor'] += int(resultado_goles[0])
        equipos[nombres_equipos[0]]['goles_contra'] += int(resultado_goles[1])
        equipos[nombres_equipos[1]]['goles_favor'] += int(resultado_goles[0])
        equipos[nombres_equipos[1]]['goles_contra'] += int(resultado_goles[1])
        if resultados[0] == 'victorias':
            equipos[nombres_equipos[0]]['victorias'] += 1
            equipos[nombres_equipos[1]]['derrotas'] += 1
        elif resultados[0] == 'derrotas':
            equipos[nombres_equipos[0]]['derrotas'] += 1
            equipos[nombres_equipos[1]]['victorias'] += 1
        else:
            equipos[nombres_equipos[0]]['empates'] += 1
            equipos[nombres_equipos[1]]['empates'] += 1
        equipos[nombres_equipos[0]]['puntos'] += puntos[0]
        equipos[nombres_equipos[1]]['puntos'] += puntos[1]

def ordena_equipos():
    equipos_ordenados = list(sorted(equipos.items(), key=lambda e: e[1]['puntos'], reverse=True))
    return equipos_ordenados

def muestra_clasificacion(equipos_ordenados):
    print('CLASIFICACIÓN FINAL'.center(100, '_')+'\n')
    print('Puesto'.ljust(8,' ') + 'Equipos'.ljust(20, ' ') + 'Goles a favor'.center(16, ' ') + 'Goles en contra'.center(16, ' ') + 'Victorias'.center(12, ' ') + 'Empates'.center(10, ' ') + 'Derrotas'.center(10, ' ') + 'Puntos'.center(10, ' ') + '\n' + ''.ljust(100,'-'))
    for num, equipo in enumerate(equipos_ordenados):
        print((str(num+1) +'.').ljust(3, ' ').center(8, ' ')+ equipo[0].ljust(20, ' ') + str(equipo[1]['goles_favor']).center(16, ' ') + str(equipo[1]['goles_contra']).center(16, ' ') + str(equipo[1]['victorias']).center(12, ' ') + str(equipo[1]['empates']).rjust(2, ' ').center(10, ' ')+ ' ' + str(equipo[1]['derrotas']).center(10, ' ') + str(equipo[1]['puntos']).center(10, ' '))

def goleador_goleado():
    mas_goleadores = list(sorted(equipos.items(), key=lambda e: e[1]['goles_favor'], reverse=True)[0])
    print(f'Equipo más goleador: {mas_goleadores[0]}, con {mas_goleadores[1]['goles_favor']} goles a favor.')

    mas_goleados = list(sorted(equipos.items(), key=lambda e: e[1]['goles_contra'], reverse=True)[0])
    print(f'Equipo más goleado: {mas_goleados[0]}, con {mas_goleados[1]['goles_contra']} goles en contra.')

def main_equipos():
    f = open('Datos/historic partits.txt', 'r')
    for linea in f:
        partido = linea.strip('\n').split('\t')
        nombres_equipos = [partido[0], partido[2]]
        resultado_goles = partido[1].split('-')
        equipos.setdefault(nombres_equipos[0], {'goles_favor':0, 'goles_contra':0, 'victorias':0, 'empates':0, 'derrotas':0, 'puntos':0})
        equipos.setdefault(nombres_equipos[1], {'goles_favor':0, 'goles_contra':0, 'victorias':0, 'empates':0, 'derrotas':0, 'puntos':0})
        resultados = define_resultado(resultado_goles)
        puntos = calcula_puntos(resultados)
        actualiza_equipo(nombres_equipos, resultado_goles, resultados, puntos)
    goleador_goleado()
    equipos_ordenados = ordena_equipos()
    muestra_clasificacion(equipos_ordenados)
    f.close()

main_equipos()

```
