```python


#  Generar contrasenyes de longitud variable.

# » Decidir si han d'incloure:
# Majúscules
# Minúscules
# Números
# Signes especials
# » Assegurar-te que com a mínim apareix un caràcter de cada tipus seleccionat.

# » Aleatoritzar completament la contrasenya.

# » Retornar la contrasenya generada.


# La funció principal del generador ha de tenir valors per defecte, exactament els següents:

# majuscules=True
# minuscules=True
# numeros=True
# signes=False
# Aquests paràmetres són obligatoris i formen part del disseny del sistema.
```

```python
import random as rd

password = []

def validar_caracter(length, types, counters, char):
    remaining = length - len(password)
    block = 0
    print(list(zip(types, counters)))
    if remaining <= (types[0]+types[1]+types[2]+types[3]):
        for type, counter in zip(types,counters):
            print(f'Type: {type}. Counter: {counter}. Char: {char}')
            if type == True and (types.index(type)+1) == char:
                if counter == 0:
                    print('hola')
                    return True
                else:
                    block = 1
            elif type != True and (types.index(type)+1) == char:
                return True
    print(f'Remaining {remaining}. Block {block}.')
    if block >= (types[0]+types[1]+types[2]+types[3]):
        return True
    else:
        return False


def genera_mayus():
    password.append(chr(rd.randint(65,90)))

def genera_minus():
    password.append(chr(rd.randint(97,122)))

def genera_num():
    password.append(chr(rd.randint(48,57)))

def genera_symbol():
    rangos = [[33,47],[58,64],[91,96],[123,126]]
    nums = []
    for rango in rangos:
        for num in range(rango[0],rango[1]):
            nums.append(num)
    simbolo = nums[rd.randint(0,len(nums)-1)]
    password.append(chr(simbolo))

def crear_contraseña(password_length, mayus=True, minus=True, nums=True, symbols=False):
    if (mayus + minus + nums + symbols) <= password_length:
        password_string = ''
        mayus_chars = 0
        minus_chars = 0
        num_chars = 0
        symbol_chars = 0
        while len(password) < password_length:
            while True:
                char = rd.randint(1, 4)
                if validar_caracter(password_length, [mayus, minus, nums, symbols], [mayus_chars, minus_chars, num_chars, symbol_chars], char) == False:
                    break
                else:
                    match char:
                        case 1:
                            if mayus==True:
                                genera_mayus()
                                mayus_chars += 1
                                break
                        case 2:
                            if minus==True:
                                genera_minus()
                                minus_chars += 1
                                break
                        case 3:
                            if nums==True:
                                genera_num()
                                num_chars += 1
                                break
                        case 4:
                            if symbols==True:
                                genera_symbol()
                                symbol_chars += 1
                                break
                    
            
        # rd.shuffle(password)
        for i in password:
            password_string += i
        password.clear()
        return password_string
    else:
        return 'Bro...'

print(crear_contraseña(3, True, True, False, True))

for i in range(10):
    print(crear_contraseña(3,True,False,False,True))
```
