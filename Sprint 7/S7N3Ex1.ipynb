```python
import random as rd

password = []

def validar_contraseña(len, mayus, minus, nums, symbols, counters):
    total = 0
    for counter in counters:
        total += counter
    remaining = len - total - mayus - minus - nums - symbols
    print(counters)
    if remaining > 0:
        if counters[0] == 0 and mayus == True:
            genera_mayus()
            return False
        if counters[1] == 0 and minus == True:
            genera_minus()
            return False
        if counters[2] == 0 and nums == True:
            genera_num()
            return False
        if counters[3] == 0 and symbols == True:
            genera_symbol()
            return False
    else:
        return True

def genera_mayus():
    password.append(chr(rd.randint(65,90)))
    return

def genera_minus():
    password.append(chr(rd.randint(97,122)))
    return

def genera_num():
    password.append(chr(rd.randint(48,57)))
    return

def genera_symbol():
    rangos = [[33,47],[58,64],[91,96],[123-126]]
    rango = rd.randint(1,4)
    password.append(chr(rd.randint(rangos[rango][0],rangos[rango][1])))
    return

def crear_contraseña(password_length, mayus=True, minus=True, nums=True, symbols=False):
    password_string = ''
    mayus_chars = 0
    minus_chars = 0
    num_chars = 0
    symbol_chars = 0
    for i in range(password_length):
        while 1==1:
            char = rd.randint(1, 4)
            if validar_contraseña(password_length, mayus, minus, nums, symbols, [mayus_chars, minus_chars, num_chars, symbol_chars]) == False:
                break
            match char:
                case 1:
                    if mayus==True:
                        genera_mayus()
                        mayus_chars += 1
                        break
                case 2:
                    if minus==True:
                        genera_minus()
                        minus_chars += 1
                        break
                case 3:
                    if nums==True:
                        genera_num()
                        num_chars += 1
                        break
                case 4:
                    if symbols==True:
                        genera_symbol()
                        symbol_chars += 1
                        break
                
        
    rd.shuffle(password)
    for i in password:
        password_string += i
    return password_string

print(crear_contraseña(4, False, False, False, True))
```
