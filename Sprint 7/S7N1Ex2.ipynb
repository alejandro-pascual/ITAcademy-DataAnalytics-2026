```python
clientes = [ 
    {"id_cliente": 1, "nombre": "Laura Serra", "edad": 23, "grupo_edad": "18-25", "codigo_comportamiento": 4}, 
    {"id_cliente": 2, "nombre": "Marc Vidal", "edad": 35, "grupo_edad": "26-40", "codigo_comportamiento": 3}, 
    {"id_cliente": 3, "nombre": "Ana López", "edad": 29, "grupo_edad": "26-40", "codigo_comportamiento": 2}, 
    {"id_cliente": 4, "nombre": "Joan Riera", "edad": 19, "grupo_edad": "18-25", "codigo_comportamiento": 2}, 
    {"id_cliente": 5, "nombre": "Mónica Pérez", "edad": 47, "grupo_edad": "41-60", "codigo_comportamiento": 4}, 
    {"id_cliente": 6, "nombre": "Luis García", "edad": 52, "grupo_edad": "41-60", "codigo_comportamiento": 1}, 
    {"id_cliente": 7, "nombre": "Pilar Sánchez", "edad": 61, "grupo_edad": "60+", "codigo_comportamiento": 3}, 
    {"id_cliente": 8, "nombre": "Eva Martín", "edad": 38, "grupo_edad": "26-40", "codigo_comportamiento": 0}, 
    {"id_cliente": 9, "nombre": "Diego Romero", "edad": 24, "grupo_edad": "18-25", "codigo_comportamiento": 1}, 
    {"id_cliente": 10, "nombre": "Núria Costa", "edad": 33, "grupo_edad": "26-40", "codigo_comportamiento": 4} 
] 

separador = '-----------------------------------------------------------------------------------------------------------------------------------------------------------------------'
```

```python
def procesar_clientes():
    clientes_procesados = []
    mensajes = [
        '¡Echa un vistazo a nuestros productos destacados y ofertas!',
        'Encuentra los productos añadidos a tu cesta en el botón en la esquina superior derecha.',
        'Ya casi lo tienes. Solo un paso más.',
        'Tus productos siguen guardados en tu cesta.',
        'Dejaste algunos productos en tu cesta de compra la última vez. Todavía puedes acceder a ellos.'
    ]
    for cliente in clientes:
        segmento_y_factor = obtener_segmento_y_factor(cliente)
        factor_edad = obtener_factor_edad(cliente)
        riesgo = calcular_riesgo(segmento_y_factor[1], factor_edad)
        cliente['segmento'] = segmento_y_factor[0]
        cliente['mensaje'] = mensajes[cliente.get('codigo_comportamiento')]
        cliente['factor_codigo'] = segmento_y_factor[1]
        cliente['factor_edad'] = factor_edad
        cliente['riesgo_final'] = riesgo
        clientes_procesados.append(cliente)
    return clientes_procesados

def obtener_segmento_y_factor(cliente):
    codigo = cliente['codigo_comportamiento']
    segmento_factor = {
        0: ['Navega pero no añade productos', 0.10],
        1: ['Añade productos pero no llega al checkout', 0.25],
        2: ['Llega al checkout pero no paga', 0.50],
        3: ['Cesta de compra abandonada hace menos de 24 horas', 0.75],
        4: ['Cesta de compra abandonada hace más de 24 horas', 0.90]
    }
    
    return segmento_factor[codigo]

def obtener_factor_edad(cliente):
    grupo_edad = cliente['grupo_edad']
    factor_edad = {
        '18-25': 1.20,
        '26-40': 1.00,
        '41-60': 0.85,
        '60+': 0.70
    }
    return factor_edad[grupo_edad]

def calcular_riesgo(factor_codigo, factor_edad):
    return factor_codigo * factor_edad

def mostrar_diccionario():
    diccionario = {}
    clientes_procesados = procesar_clientes()
    for item in clientes_procesados:
        diccionario[item['id_cliente']] = item
        print(item)
    return diccionario

def mostrar_respuestas(diccionario):
    print('¿Qué clientes se han de contactar primero en una campaña de retargetting?')
    clientes_prioritarios = []
    for cliente in diccionario:
        if diccionario[cliente]['riesgo_final'] >= 0.63:
            clientes_prioritarios.append([diccionario[cliente]['nombre'],diccionario[cliente]['riesgo_final']])
    clientes_prioritarios.sort(key=lambda cli: cli[1], reverse=True)
    for cliente_prioritario in clientes_prioritarios:
        print(f'{clientes_prioritarios.index(cliente_prioritario)+1}) {cliente_prioritario[0]}. Riesgo final: {cliente_prioritario[1]}')
        
    print('\n¿Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes? (Por ejemplo: ¿un cliente joven que abandonó la cesta de compra hace 24 horas tiene el mismo riesgo que uno de más edad?)')
    clientes_4_por_edad = []
    for cliente in diccionario:
        if diccionario[cliente]['codigo_comportamiento'] == 4:
            clientes_4_por_edad.append([diccionario[cliente]['nombre'],diccionario[cliente]['edad'],diccionario[cliente]['segmento'],diccionario[cliente]['riesgo_final']])
    clientes_4_por_edad.sort(key=lambda cli: cli[1], reverse=True)
    for cliente_4_por_edad in clientes_4_por_edad:
        print(f'{cliente_4_por_edad[0]}. Cliente de {cliente_4_por_edad[1]} años de edad. {cliente_4_por_edad[2]}. Riesgo final: {cliente_4_por_edad[3]}')
    
    print('\n¿Qué clientes presentan tan poco interés que no merece la pena invertirle presupuesto?')
    clientes_ignorables = []
    for cliente in diccionario:
        if diccionario[cliente]['riesgo_final'] <= 0.3:
            clientes_ignorables.append([diccionario[cliente]['nombre'],diccionario[cliente]['riesgo_final']])
    clientes_ignorables.sort(key=lambda cli: cli[1])
    for cliente_ignorable in clientes_ignorables:
        print(f'{clientes_ignorables.index(cliente_ignorable)+1}) {cliente_ignorable[0]}. Riesgo final: {cliente_ignorable[1]}')
        
    print('\n¿Qué cliente sería el más rentable de recuperar?')
    clientes_rentables = []
    for cliente in diccionario:
        if diccionario[cliente]['codigo_comportamiento'] == 4:
            clientes_rentables.append([diccionario[cliente]['nombre'],diccionario[cliente]['codigo_comportamiento'],diccionario[cliente]['riesgo_final']])
    clientes_rentables.sort(key=lambda cli: cli[2], reverse=True)
    print(f'{clientes_rentables[0][0]} con código de comportamiento {clientes_rentables[0][1]}. Riesgo final: {clientes_rentables[0][2]}')
    
    return

def main_marketing(): 
    print('Análisis de preguntas de marketing\n'+separador)
    diccionario = mostrar_diccionario()
    print(separador)
    mostrar_respuestas(diccionario)
    
```

```python
main_marketing()
```
