### Funciones

**validar_datos:** Recibe el registro a validar del que se validarán los datos y qué tipo de validaciones deberá realizar sobre ellos en su respectivo orden. Devuelve True si el registro pasa correctamente todas las validaciones. De lo contrario, devuelve False.

**calcular_puntuacion:** Recibe los valores de los distintos campos a partir de los cuales se calcula la puntuación y realiza una operación de la que devuelve su resultado.

**clasificar_candidato:** Devuelve una clasificación según el intervalo donde pertenezca la puntuación recibida por parámetro.

**mostrar_individual:** Imprime un mensaje a partir del nombre y la puntuación del registro.

**obtener_datos:** Recoge los datos desde la variable inputs_test, si se ha seleccionado el modo en conjunto, o pedirá los datos uno por uno, en modo individual.

**mostrar_informe:** Muestra nombre, teléfono, puntuación y clasificación de todos los candidatos en la lista.

**main_seleccion:** Inicia todo el programa preguntando al usuario cómo quiere introducir los datos, siempre que haya menos de 10 registros en la lista de candidatos.

```python
import re

candidatos = []
separador = '-----------------------------------------------------------------------------------------------------------------'

inputs_test = [
    "Laura Serra", "608112233", 6, 3, 3, 3,
    "Clara Gómez", "655221199", 3, 3, 3, 2,
    "Sergio López", "644112211", 3, 2, 2, 2,
    "Candidat Error", "611000000", 5, 7, 3, 2, # Catalán = 7 → NO VÁLIDO
    "Marta Ruiz", "699887766", 2, 3, 2, 1,
    "Marc Vidal", "612443322", 6, 3, 3, 3,
    "Júlia Romero", "677998877", 5, 3, 3, 2,
    "Pol Navarro", "699554433", 4, 3, 2, 2,
    "David Ferrer", "677665544", 2, 1, 1, 1,
    "Anna Paredes", "633221144", 1, 2, 1, 0,
    "Anibal Paredes", "644272727", 1, 2, 1, 0,
    "Núria Casals", "600112200", 8, 2, 3, 3 # 11ª candidata
]

def validar_datos(registro, validaciones):
    if registro != None:
        for campo in range(len(validaciones)):
            match validaciones[campo]:
                case 'smalltext':
                    if re.search('[^\\sa-zA-ZÀ-ÿ]',registro[campo]) != None:
                        return False
                case 'int':
                    try:
                        if type(int(registro[campo])) != int:
                            return False
                    except:
                        return False
                case 'opt3':
                    try:
                        if type(int(registro[campo])) != int or int(registro[campo]) > 3 or int(registro[campo]) < 0:
                            return False
                    except:
                        print(type(int(registro[campo])), registro[campo])
                        return False

        return True
    else:
        return False

def calcular_puntuacion(exp, cat, excel, titulo):
    return (int(exp)*1.5)+int(cat)+int(excel)+int(titulo)

def clasificar_candidato(puntuacion):
    match puntuacion:
        case _ if puntuacion < 8:
            return 'No prioritario'
        case _ if puntuacion < 12:
            return 'Candidato adecuado'
        case _ if puntuacion >= 12:
            return 'Prioridad alta'

def mostrar_individual(nombre, puntuacion):
    print('El/La candidato/a '+str(nombre)+' ha obtenido un total de '+str(puntuacion)+' puntos.')

def obtener_datos(modo):
    if modo == 0:
        nombre = input('Introduce nombre y apellido:')
        tel = input('Introduce número de teléfono:')
        exp = input('Introduce años de experiencia:')
        cat = input('Introduce el nivel de conocimiento de catalán:\n(0) Sin conocimiento\n(1) Hasta nivel B1\n(2) Nivel B2\n(3) C1 o superior')
        excel = input('Introduce el nivel de dominio de Excel:\n(0) Nivel básico\n(1) Uso de funciones\n(2) Tablas dinámicas\n(3) Visual Basic y macros')
        titulo = input('Introduce el título académico:\n(0) ESO\n(1) FP\n(2) Universidad\n(3) Máster o superior')
        fila = [nombre,tel,exp,cat,excel,titulo]
        if validar_datos(fila, ['smalltext','int','int','opt3','opt3','opt3']):
            puntuacion = calcular_puntuacion(fila[2],fila[3],fila[4],fila[5])
            clasificacion = clasificar_candidato(puntuacion)
            mostrar_individual(fila[0],puntuacion)
            candidatos.append([fila[0],fila[1], puntuacion, clasificacion])
        else:
            print('Datos incorrectos. El registro no ha sido añadido.')
    elif modo == 1:
        for i in range(0, len(inputs_test), 6):
            fila = []
            for j in range(i, i+6):
                fila.append(str(inputs_test[j]))
            if validar_datos(fila, ['smalltext','int','int','opt3','opt3','opt3']):
                puntuacion = calcular_puntuacion(fila[2],fila[3],fila[4],fila[5])
                clasificacion = clasificar_candidato(puntuacion)
                mostrar_individual(fila[0],puntuacion)
                candidatos.append([fila[0],fila[1], puntuacion, clasificacion])
                if len(candidatos) >= 10:
                    break

            else:
                print('Datos incorrectos en el registro número '+str(int(i/6)+1)+'. Saltando dicho registro...')

def mostrar_informe():
    candidatos.sort(key=lambda cdt: cdt[2], reverse=True)
    for candidato in candidatos:
        print('Nombre: '+candidato[0]+' | Teléfono: '+candidato[1]+' | Puntuación: '+str(candidato[2])+' | Clasificación: '+candidato[3])

def main_seleccion():
    print('Herramienta de selección de candidatos')
    while len(candidatos) < 10:
        print(separador)
        modo = int(input('¿Quieres introducir los datos individualmente o en conjunto?\n(0) Individual\n(1) Conjunto\n'))
        if modo == 0:
            print('Introducción de candidato número',len(candidatos)+1)
        elif modo == 1:
            print('Introducción de candidatos')

        obtener_datos(modo)
    print(separador+'\nSe ha alcanzado el máximo de 10 candidatos registrados.\n'+separador)
    mostrar_informe()

main_seleccion()
```

### Funciones

**convertir_pacientes:** De la lista pacientes, extrae uno por uno los registros, los secciona y los coloca en un diccionario.

**transformar_sintomas:** Separa y realiza varias correcciones en los síntomas que se le pasan por parámetro, para poder manejarlos mejor.

**definir_riesgo_sintomas:** Asocia los síntomas del paciente recibidos por parámetro, con el nivel de riesgo correspondiente de la lista niveles_riesgo. Entonces, devuelve dicho número.

**calcular_riesgo_paciente:** Calcula el total de nivel de riesgo del paciente y lo devuelve.

**ordenar_pacientes:** Crea y devuelve un diccionario ordenado a partir del diccionario de pacientes.

**mostrar_pacientes:** Imprime los pacientes de uno en uno.

**main_pacientes:** Inicia todo el programa llamando a las funciones necesarias.

```python
separador = '-----------------------------------------------------------------------------------------------------------------------------------------------------------------------'

pacientes = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

niveles_riesgo = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}

def convertir_pacientes():
    dic_pacientes = {}
    for paciente in pacientes:
        datos_paciente = paciente.split('|')
        sintomas = transformar_sintomas(datos_paciente[2])
        dic_pacientes[pacientes.index(paciente)] = {
            'nombre':datos_paciente[0],
            'edad':datos_paciente[1],
            'sintomas':sintomas,
            'riesgo_total': calcular_riesgo_paciente(definir_riesgo_sintomas(sintomas))
            }
    return dic_pacientes

def transformar_sintomas(sintomas):
    sintomas_separados = sintomas.split(',')
    sintomas_normalizados = []
    for sintoma in sintomas_separados:
        sintomas_normalizados.append(sintoma.strip(' ').lower().replace(' ','_'))
    return sintomas_normalizados

def definir_riesgo_sintomas(sintomas):
    riesgo = []
    for sintoma in sintomas:
        if sintoma in niveles_riesgo:
            riesgo.append(niveles_riesgo[sintoma])
    return riesgo

def calcular_riesgo_paciente(nivel_riesgo):
    riesgo_total = 0
    for riesgo in nivel_riesgo:
        riesgo_total += riesgo
    return riesgo_total

def ordenar_pacientes(dic_pacientes):
    pacientes_ordenados = dict(sorted(dic_pacientes.items(), key=lambda i: i[1]['riesgo_total'], reverse=True))
    return pacientes_ordenados

def mostrar_pacientes(pacientes):
    for paciente in ordenar_pacientes(pacientes).values():
        print(paciente)

def main_pacientes():
    print('Indicador de riesgo de pacientes\n'+separador)
    pacientes = convertir_pacientes()
    mostrar_pacientes(pacientes)
    print(separador)

main_pacientes()
```

### Funciones

**clasificar_seguros:** Devuelve los datos asociados al nivel del seguro recibido por parámetro.

**calcular_precio_final:** Calcula y devuelve el precio final a partir del precio base y la edad.

**agregar_clientes:** De la lista clientes, extrae uno por uno los registros, los secciona y los coloca en un diccionario.

**ordenar_seguros:** Devuelve un diccionario nuevo creado a partir del diccionario de seguros, ordenado por probabilidad de fraude (1) y precio final (2), en esa prioridad de orden.

**mostrar_clientes:** Imprime todos los clientes uno por uno.

**construir_informe:** Calcula todos los datos necesarios para el informe y los guarda en un string que constituye el informe.

**mostrar_informe:** Muestra el informe recibido por la función construir_informe.

**main_seguro:** Inicia todo el programa llamando a las funciones necesarias.

```python
separador = '-----------------------------------------------------------------------------------------------------------------------------------------------------------------------'


clientes = [
    {"nombre": "Ana", "edad": 42, "codigo": 0},
    {"nombre": "Carlos", "edad": 30, "codigo": 2},
    {"nombre": "Isabel", "edad": 55, "codigo": 3},
    {"nombre": "Jorge", "edad": 40, "codigo": 1},
    {"nombre": "Marta", "edad": 28, "codigo": 0}
]


niveles_seguro = {
    0: {"categoria": "Bajo riesgo", "precio_base": 120, "fraude": 0.01},
    1: {"categoria": "Riesgo medio", "precio_base": 200, "fraude": 0.05},
    2: {"categoria": "Riesgo alto", "precio_base": 350, "fraude": 0.15},
    3: {"categoria": "Riesgo crítico", "precio_base": 500, "fraude": 0.30}
}


def clasificar_seguros(codigo):
    return niveles_seguro[codigo]


def calcular_precio_final(base_edad):
    return round(int(base_edad[0]) * (1+base_edad[1]/100), 2)


def agregar_clientes():
    seguros = {}
    for cliente in clientes:
        datos_clasificacion = clasificar_seguros(cliente['codigo'])
        seguros[clientes.index(cliente)] = {
            'nombre': cliente['nombre'],
            'edad': cliente['edad'],
            'categoria': datos_clasificacion['categoria'],
            'precio_base': datos_clasificacion['precio_base'],
            'precio_final': calcular_precio_final([datos_clasificacion['precio_base'],cliente['edad']]),
            'probabilidad_fraude': datos_clasificacion['fraude']

        }
    return seguros


def ordenar_seguros(seguros):
    seguros_ordenados = dict(sorted(seguros.items(), key=lambda c: c[1]['precio_final'], reverse=True))
    seguros_ordenados = dict(sorted(seguros_ordenados.items(), key=lambda c: c[1]['probabilidad_fraude'], reverse=True))
    return seguros_ordenados


def mostrar_clientes(seguros):
    for seguro in seguros.values():
        print(seguro)
    return


def construir_informe(seguros):
    fraudulento = None
    caro = None
    riesgo_prioritario = []
    max_fraude = 0
    max_precio = 0
    for seguro in seguros:
        cliente = seguros[seguro]
        if cliente['probabilidad_fraude'] > max_fraude:
            max_fraude = cliente['probabilidad_fraude']
        if cliente['precio_final'] > max_precio:
            max_precio = cliente['precio_final']
        if cliente['probabilidad_fraude'] == max_fraude:
            fraudulento = cliente
        if cliente['precio_final'] == max_precio:
            caro = cliente
        if cliente['categoria'] == 'Riesgo crítico' or cliente['categoria'] == 'Riesgo alto':
            riesgo_prioritario.append(cliente)
    informe = f'El cliente con más probabilidad de fraude es: {fraudulento['nombre']}, con una probabilidad del {max_fraude}.\nEl cliente que paga un precio final más alto es: {caro['nombre']}, con un precio final de {max_precio}.\n{f'Los clientes prioritarios de revisión manual son: ' if len(riesgo_prioritario) > 1 else 'El cliente prioritario de revisión manual es '}'
    for arriesgado in riesgo_prioritario:
        if riesgo_prioritario.index(arriesgado) < len(riesgo_prioritario)-1:
            informe += f'{arriesgado['nombre']}, '
        else:
            informe += f'{arriesgado['nombre']}.'
    return informe


def mostrar_informe(seguros):
    informe = construir_informe(seguros)
    print(informe)


def main_seguro():
    print('Información de seguro\n'+separador)
    seguros = agregar_clientes()
    seguros_ordenados = ordenar_seguros(seguros)
    mostrar_clientes(seguros_ordenados)
    print(separador)
    mostrar_informe(seguros_ordenados)

main_seguro()
```

Por último, reordena aleatoriamente todos los caracteres de la contraseña y devuelve el resultado.

```python
import random as rd

password = []

def validar_caracter(length, types, counters, char):
    remaining = length - len(password)
    block = 0
    if remaining <= (types[0]+types[1]+types[2]+types[3]):
        for i, (type, counter) in enumerate(zip(types,counters)):
            if i+1 == char:
                if type == True:
                    if counter == 0:
                        return True
                elif type != True:
                    return True
            if type == True and counter == 0:
                block += 1
    if block == 0:
        return True
    else:
        return False


def genera_mayus():
    password.append(chr(rd.randint(65,90)))

def genera_minus():
    password.append(chr(rd.randint(97,122)))

def genera_num():
    password.append(chr(rd.randint(48,57)))

def genera_symbol():
    rangos = [[33,47],[58,64],[91,96],[123,126]]
    nums = []
    for rango in rangos:
        for num in range(rango[0],rango[1]):
            nums.append(num)
    simbolo = nums[rd.randint(0,len(nums)-1)]
    password.append(chr(simbolo))

def crear_contraseña(password_length, mayus=True, minus=True, nums=True, symbols=False):
    if (mayus + minus + nums + symbols) <= password_length:
        password_string = ''
        mayus_chars = 0
        minus_chars = 0
        num_chars = 0
        symbol_chars = 0
        while len(password) < password_length:
            while True:
                char = rd.randint(1, 4)
                if validar_caracter(password_length, [mayus, minus, nums, symbols], [mayus_chars, minus_chars, num_chars, symbol_chars], char) == False:
                    break
                else:
                    match char:
                        case 1:
                            if mayus==True:
                                genera_mayus()
                                mayus_chars += 1
                                break
                        case 2:
                            if minus==True:
                                genera_minus()
                                minus_chars += 1
                                break
                        case 3:
                            if nums==True:
                                genera_num()
                                num_chars += 1
                                break
                        case 4:
                            if symbols==True:
                                genera_symbol()
                                symbol_chars += 1
                                break

        rd.shuffle(password)
        for i in password:
            password_string += i
        password.clear()
        return password_string
    else:
        return 'No puedo meter más tipos de caracteres, que caracteres tiene la contraseña.'


print(crear_contraseña(12, True, True, True, True))
```

```python
ruta = '/content/drive/My Drive/Data Analytics/historic partits.txt'

equipos = {}

def define_resultado(goles):
    if goles[0] > goles[1]:
        return ['victorias','derrotas']
    if goles[0] < goles[1]:
        return ['derrotas','victorias']
    if goles[0] == goles[1]:
        return ['empates','empates']

def calcula_puntos(resultados):
    puntos = [0, 0]
    for i in range(2):
        if resultados[i] == 'victorias':
            puntos[i] = 3
        elif resultados[i] == 'empates':
            puntos[i] = 1
        elif resultados[i] == 'derrotas':
            puntos[i] = 0
    return puntos

def actualiza_equipo(nombres_equipos, resultado_goles, resultados, puntos):
        equipos[nombres_equipos[0]]['goles_favor'] += int(resultado_goles[0])
        equipos[nombres_equipos[0]]['goles_contra'] += int(resultado_goles[1])
        equipos[nombres_equipos[1]]['goles_favor'] += int(resultado_goles[0])
        equipos[nombres_equipos[1]]['goles_contra'] += int(resultado_goles[1])
        if resultados[0] == 'victorias':
            equipos[nombres_equipos[0]]['victorias'] += 1
            equipos[nombres_equipos[1]]['derrotas'] += 1
        elif resultados[0] == 'derrotas':
            equipos[nombres_equipos[0]]['derrotas'] += 1
            equipos[nombres_equipos[1]]['victorias'] += 1
        else:
            equipos[nombres_equipos[0]]['empates'] += 1
            equipos[nombres_equipos[1]]['empates'] += 1
        equipos[nombres_equipos[0]]['puntos'] += puntos[0]
        equipos[nombres_equipos[1]]['puntos'] += puntos[1]

def ordena_equipos():
    equipos_ordenados = list(sorted(equipos.items(), key=lambda e: e[1]['puntos'], reverse=True))
    return equipos_ordenados

def muestra_clasificacion(equipos_ordenados):
    print('CLASIFICACIÓN FINAL'.center(100, '_')+'\n')
    print('Puesto'.ljust(8,' ') + 'Equipos'.ljust(20, ' ') + 'Goles a favor'.center(16, ' ') + 'Goles en contra'.center(16, ' ') + 'Victorias'.center(12, ' ') + 'Empates'.center(10, ' ') + 'Derrotas'.center(10, ' ') + 'Puntos'.center(10, ' ') + '\n' + ''.ljust(100,'-'))
    for num, equipo in enumerate(equipos_ordenados):
        print((str(num+1) +'.').ljust(3, ' ').center(8, ' ')+ equipo[0].ljust(20, ' ') + str(equipo[1]['goles_favor']).center(16, ' ') + str(equipo[1]['goles_contra']).center(16, ' ') + str(equipo[1]['victorias']).center(12, ' ') + str(equipo[1]['empates']).rjust(2, ' ').center(10, ' ')+ ' ' + str(equipo[1]['derrotas']).center(10, ' ') + str(equipo[1]['puntos']).center(10, ' '))

def goleador_goleado():
    mas_goleadores = list(sorted(equipos.items(), key=lambda e: e[1]['goles_favor'], reverse=True)[0])
    print(f'Equipo más goleador: {mas_goleadores[0]}, con {mas_goleadores[1]['goles_favor']} goles a favor.')

    mas_goleados = list(sorted(equipos.items(), key=lambda e: e[1]['goles_contra'], reverse=True)[0])
    print(f'Equipo más goleado: {mas_goleados[0]}, con {mas_goleados[1]['goles_contra']} goles en contra.')

def main_equipos():
    f = open(ruta, 'r')
    for linea in f:
        partido = linea.strip('\n').split('\t')
        nombres_equipos = [partido[0], partido[2]]
        resultado_goles = partido[1].split('-')
        equipos.setdefault(nombres_equipos[0], {'goles_favor':0, 'goles_contra':0, 'victorias':0, 'empates':0, 'derrotas':0, 'puntos':0})
        equipos.setdefault(nombres_equipos[1], {'goles_favor':0, 'goles_contra':0, 'victorias':0, 'empates':0, 'derrotas':0, 'puntos':0})
        resultados = define_resultado(resultado_goles)
        puntos = calcula_puntos(resultados)
        actualiza_equipo(nombres_equipos, resultado_goles, resultados, puntos)
    goleador_goleado()
    equipos_ordenados = ordena_equipos()
    muestra_clasificacion(equipos_ordenados)
    f.close()

main_equipos()

```
